```python
from ultralytics import YOLO

# Tải cấu hình trọng số bản Small đúng như kế hoạch
model = YOLO('yolov8s.pt')

# Kích hoạt lệnh train trỏ thẳng vào file data.yaml trên Drive của bạn
results = model.train(
    data='/content/drive/MyDrive/Traffic_Sign_Project/data/dataset/data.yaml',
    epochs=100,
    batch=16,
    imgsz=640,
    device=0
)
```
